Let's create a function that check for missing data in the two datasets (train and test).


```python
# Write and run the above code in the following code section.
def missing_data(data):
    total = data.isnull().sum().sort_values(ascending = False)
    percent = (data.isnull().sum()/data.isnull().count()*100).sort_values(ascending = False)
    return pd.concat([total,percent], axis=1, keys=['Total', 'Percent'])

missing_data(train)
```

# <a id='preprocess'>Data preprocessing</a>  


Saya membuat Feature Engineering dari Informasi Atribut sebelumnya. Lalu, seperti yang diketahui Variabel lesion merupakan kode medis yang terdiri dari beberapa digit dengan makna berbeda. Untuk menyederhanakan kompleksitas(mengurangi dimensionalitas dan menghindari sparsity) yang nantinya menggunakan onehot encoding, saya hanya memakai digit kedua (type of lesion) yang digunakan karena merepresentasikan jenis kondisi utama. Dari tipe tersebut, dilakukan penyederhanaan lebih lanjut dengan hanya mempertahankan kondisi strangulation (kode 2), karena memiliki tingkat keparahan paling tinggi serta relevansi yang kuat terhadap outcome pasien. Sementara itu, tipe kondisi lainnya digabungkan ke dalam satu kategori. sehingga hasilnya akan ditransformasikan menjadi variabel biner, yaitu : stranglutation = 1 dan other = 0. 

```python
def extract_lesion_features(df, col_name):

    s = df[col_name].astype(str).str.zfill(5) 

    df[f'{col_name}_type'] = ((df[col_name] // 1000) % 10)     

    df[f'{col_name}_is_strangulation'] = (df[f'{col_name}_type'] == 2).astype(int)
    return df
```

```python
def feature_engineering(df):
    df = df.copy()

    # atur lesion code
    for col in ['lesion_1', 'lesion_2', 'lesion_3']:
        df = extract_lesion_features(df, col)

    df['has_strangulation'] = (
        df['lesion_1_is_strangulation'] |
        df['lesion_2_is_strangulation'] |
        df['lesion_3_is_strangulation']
    ).astype(int)

    # syarat
    df['temp_deviation'] = (df['rectal_temp'] - 37.8).abs()

    # ubah kategori yg berbeda dari informasi
    df['capillary_refill_time'] = df['capillary_refill_time'].replace({'3': 'more_3_sec'})
    df['nasogastric_reflux'] = df['nasogastric_reflux'].replace({'slight': 'less_1_liter'})
    df['peristalsis'] = df['peristalsis'].replace({'distend_small': 'hypomotile'})
    df['rectal_exam_feces'] = df['rectal_exam_feces'].replace({'serosanguious': 'absent'})

    # binary encoding
    df['surgery'] = (df['surgery'] == 'yes').astype(int)
    df['surgical_lesion'] = (df['surgical_lesion'] == 'yes').astype(int)
    df['age'] = (df['age'] == 'adult').astype(int)

    # binning
    df['pulse_severity'] = pd.cut(df['pulse'], bins=[0, 29, 40, 60, 200], labels=[0, 1, 2, 3], include_lowest=True).astype(float)
    df['protein_severity'] = pd.cut(df['total_protein'], bins=[0, 6, 7.5, 10, 100], labels=[0, 1, 2, 3], include_lowest=True).astype(float)

    # fe
    df['pulse_pcv_risk'] = df['pulse_severity'] * df['packed_cell_volume'] / 100
    df['pain_and_surgery'] = df['surgery'] * df['surgical_lesion'] 
    df['dehydration_index'] = df['packed_cell_volume'] * df['total_protein']
    df['critical_state'] = ((df['pain'] == 'depressed') & (df['peristalsis'] == 'absent')).astype(int)
    df['vital_stress'] = df['pulse'] * df['rectal_temp']
    
    # Referensi: PMC11415896
    df['gut_failure'] = (
        (df['peristalsis'] == 'absent') |
        (df['nasogastric_reflux'] == 'more_1_liter')
    ).astype(int)
    
    df['perfusion_failure'] = (
        (df['capillary_refill_time'] == 'more_3_sec') &
        (df['mucous_membrane'].isin(['pale_cyanotic', 'dark_cyanotic']))
        
    ).astype(int)
    # DROP
    df.drop(columns=['cp_data','lesion_1', 'lesion_2', 'lesion_3', 'id', 'hospital_number','rectal_temp','pulse','total_protein',
                    'respiratory_rate','lesion_1_type', 'lesion_2_type', 'lesion_3_type','lesion_1_is_strangulation', 'lesion_2_is_strangulation', 'lesion_3_is_strangulation'], inplace=True, errors='ignore')
    return df
```

## <a id='impute'>Imputation of missing data</a>  

We will use the most frequent value to impute all columns with missing data.  

```python
# Mapping ordinal
ordinal_mappings = {
    'temp_of_extremities': ['normal', 'warm', 'cool', 'cold'],
    'abdominal_distention': ['none', 'slight', 'moderate', 'severe'],
    'capillary_refill_time': ['less_3_sec', 'more_3_sec'],
    'peristalsis': ['hypermotile', 'normal', 'hypomotile', 'absent'],
    'nasogastric_tube': ['none', 'slight', 'significant'],
    'nasogastric_reflux': ['none', 'more_1_liter','less_1_liter'],
    'rectal_exam_feces': ['normal', 'increased', 'decreased', 'absent'],
    'abdomo_appearance': ['clear', 'cloudy', 'serosanguious']
}

all_mappings = {
    'peripheral_pulse': {
        'normal': 0, 'increased': 0, 'reduced': 1, 'absent': 1
    },
    'mucous_membrane': {
        'normal_pink': 0, 'bright_pink': 0, 'pale_pink': 1, 
         'pale_cyanotic': 2, 'dark_cyanotic': 2,'bright_red': 3
    }
}
```

```python
# Write and run the above code in the following code section.
from sklearn.impute import SimpleImputer
import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import FunctionTransformer
from sklearn import set_config

cat_columns = X_train.select_dtypes(include=['object', 'category']).columns.tolist()
num_cols = X_train.select_dtypes(include=['float64','int64']).columns.tolist()
ordinal_cols = list(all_mappings.keys()) + list(ordinal_mappings.keys())
cat_cols = [col for col in cat_columns if col not in ordinal_cols]

set_config(transform_output="pandas")

def custom_mapper(df):
    df_copy = df.copy()
    # Terapkan all_mappings (Mapping dictionary)
    for col, mapping in all_mappings.items():
        if col in df_copy.columns:
            df_copy[col] = df_copy[col].map(mapping)
    
    # Terapkan ordinal_mappings (Berdasarkan index list)
    for col, urutan in ordinal_mappings.items():
        if col in df_copy.columns:
            # Mengubah list jadi dictionary index: {'normal': 0, 'warm': 1, ...}
            map_index = {val: i for i, val in enumerate(urutan)}
            df_copy[col] = df_copy[col].map(map_index)
            
    return df_copy                
```

```python
# Pipeline khusus untuk kolom Ordinal (Imputasi -> Map ke Angka)
ordinal_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('mapper', FunctionTransformer(custom_mapper))
])

# Pipeline untuk kolom Nominal (Imputasi -> OneHot)
cat_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))])


#Gabungkan semuanya
preprocessor = ColumnTransformer(
    transformers=[
        ('num', SimpleImputer(strategy='median'), num_cols),
        ('ord', ordinal_transformer, ordinal_cols),
        ('nom', cat_transformer, cat_cols)
    ]
)


X_train_transformed = preprocessor.fit_transform(X_train)
X_val_transformed   = preprocessor.transform(X_val)
X_test_transformed  = preprocessor.transform(test_fe)
```

```python
train_l = X_train_transformed
val_l   = X_val_transformed
test_l  = X_test_transformed
```

```python
# definisikan sample weight agar pembobotan tidak fokus ke mayo
class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(y_train_encoded),
    y=y_train_encoded
)

sample_weights = np.array([class_weights[y] for y in y_train_encoded])

# Melakukan tuning hyperparameter agar mendapat hasil model yang baik
def objective_xgb(trial):
    params = {
    "n_estimators": trial.suggest_int("n_estimators", 100, 300),
    "max_depth": trial.suggest_int("max_depth", 3, 5),  # 🔥 lebih kecil
    "learning_rate": trial.suggest_float("learning_rate", 0.01, 0.1),  # 🔥 lebih kecil
    "subsample": trial.suggest_float("subsample", 0.6, 0.9),
    "colsample_bytree": trial.suggest_float("colsample_bytree", 0.6, 0.9),
    "min_child_weight": trial.suggest_int("min_child_weight", 3, 10),
    # 🔥 TAMBAHAN PENTING
    "reg_alpha": trial.suggest_float("reg_alpha", 0.1, 5.0, log=True),
    "reg_lambda": trial.suggest_float("reg_lambda", 1.0, 15.0, log=True),
    "gamma": trial.suggest_float("gamma", 0.5, 3.0),

    "objective": "multi:softmax",
    "num_class": 3,
    "random_state": 42,
    "n_jobs": -1
}

    skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    f1_scores = []

    for train_idx, val_idx in skf.split(X_train, y_train_encoded):
        # 1. Split data berdasarkan index fold
        X_tr, X_valid = X_train.iloc[train_idx].copy(), X_train.iloc[val_idx].copy()
        y_tr, y_valid = y_train_encoded[train_idx], y_train_encoded[val_idx]

        preprocessor_fold = clone(preprocessor)

        smple = compute_sample_weight('balanced', y_tr)
    
        # 6. Gabungkan Numerik + Kategorikal (Hasil Akhir)
        X_tr_final = preprocessor_fold.fit_transform(X_tr)
        X_valid_final = preprocessor_fold.transform(X_valid)

        X_tr_selected = X_tr_final[selected_cols]
        X_valid_selected = X_valid_final[selected_cols]

        model = XGBClassifier(**params,early_stopping_rounds=30,eval_metric="mlogloss")
        model.fit(X_tr_selected, y_tr,eval_set=[(X_valid_selected, y_valid)],
            verbose=False, sample_weight=smple)

        y_pred = model.predict(X_valid_selected)
        f1_scores.append(f1_score(y_valid, y_pred, average='macro'))

    return np.mean(f1_scores)
```
